## Question 1: Is Unemployment in August Always Higher Than in February Across Years?


### We aim to analyze whether there is a seasonal pattern in unemployment rates in Indonesia by comparing two periods: February and August for each year.

In context, many fresh graduates in Indonesia complete their studies around June–July. This leads to a potential increase in unemployment in August, as they enter the labor market but have not yet secured jobs. Therefore, August may reflect a temporary spike in unemployment due to the inflow of new job seekers.

```python
import pandas as pd
df = pd.read_csv('Data_Cleaning_Pengangguran.csv')
```

```python
df.head(2)
```

```python
df.info()
df.isnull().sum()
```

```python
df_filtered = df[df['Bulan'].isin(['Februari', 'Agustus'])]
df_filtered.head()
```

```python
df_trend = df_filtered.groupby(['Periode', 'Bulan'])['Total'].mean().reset_index()
df_trend
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")
plt.figure(figsize=(10,5))

sns.lineplot(data=df_trend, x='Periode', y='Total', hue='Bulan', marker="o")

plt.title("Unemployment Trend: February vs August", fontweight='bold')
plt.xlabel("Year")
plt.ylabel("Total Unemployment")

plt.tight_layout()
plt.show()
```

## Unemployment Trend Analysis (February vs. August)

Based on the data visualization from 2006 to 2022, the empirical evidence demonstrates that unemployment rates in August are **not always** higher than in February. This finding directly refutes the common assumption that the academic graduation cycle (schools and universities) in the middle of the year consistently triggers an immediate spike in national unemployment every August.

### Historical Overview and Trend Dynamics
A deeper look into the data reveals that the trend between these two periods is highly fluctuating and intersects over the years. From 2006 to 2010, the data shows that unemployment rates in February were actually consistently higher than in August. 

This pattern began to shift during specific years, such as 2011, 2015, and the period from 2018 to 2021, where August numbers reversed and peaked. The most drastic and extreme spike occurred in August 2020, which was a direct consequence of the macroeconomic shock caused by the COVID-19 pandemic.

### Macroeconomic Implications
In conclusion, the dynamics of Indonesia's labor market are not solely driven by seasonal factors like the influx of fresh graduates. National unemployment rates prove to be far more sensitive to broader macroeconomic conditions, quarterly job creation momentum, and external shocks occurring within that specific year.

## Question 2: Which Education Level Has the Fastest Unemployment Growth Rate?

### We aim to identify which education group has the fastest increase in unemployment over the last 5 to 10 years.

Even if a group currently has a lower total number of unemployed people, a fast growth rate can indicate a future risk. For example, university graduates may have lower unemployment today, but if their unemployment is increasing rapidly each year, it could signal a growing mismatch between education and labor market demand.

This analysis helps us identify not only current problems but also potential future risks in the labor market.

```python
import pandas as pd

df = pd.read_csv("Data_Cleaning_Pengangguran.csv")
```

```python
import pandas as pd

df = pd.read_csv('Data_Cleaning_Pengangguran.csv')
```

```python
df_yearly = df.groupby('Periode')[[
    'SLTA Umum/SMU',
    'SLTA Kejuruan/SMK',
    'Universitas'
]].mean().reset_index()
```

```python
df_yearly.tail()
```

```python
df_growth = df_yearly.set_index('Periode').pct_change() * 100

df_growth = df_growth.reset_index()

df_growth
```

```python
growth_avg = df_growth.drop(columns=['Periode']).mean()

growth_avg
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
growth_avg = df_growth.drop(columns=['Periode']).mean()
```

```python
plt.figure(figsize=(8,5))

sns.set_theme(style="whitegrid")

sns.barplot(
    x=growth_avg.index,
    y=growth_avg.values,
    palette='Oranges_r'
)

plt.title("Average Annual Unemployment Growth Rate (2006-2022)")
plt.xlabel("Education Level")
plt.ylabel("Average Growth Rate (%)")

plt.show()
```

### **Question 2: Which Education Level Has the Fastest Unemployment Growth Rate?**

Based on the labor market data analysis from 2006 to 2022, **University graduates** experienced the fastest unemployment growth rate in Indonesia, surging by an average of **5.85% annually**. This rate is significantly higher than that of **Vocational High Sch
ool (SMK)** graduates at 2.70%, while **General High School (SMU)** graduates actually showed a declining trend of **-0.72%**.

The rapid growth among university graduates reveals an academic paradox of **structural mismatch** in the labor market. The supply of new higher education graduates is expanding faster than the creation of high-skilled jobs, driving a critical gap between academic competencies and modern industrial demands. 

Meanwhile, the decline in SMU unemployment suggests that general high school graduates are either pivoting into higher education—subsequently shifting into the university category—or demonstrating greater flexibility by entering the informal sector. Historically, this surge in highly educated unemployment has been highly sensitive to macroeconomic shifts, as evidenced by two massive volatility spikes in 2015 (+36.43%) and 2018 (+25.96%). These findings strongly underscore the urgent need to align higher education curricula with real-world industry requirements.

## Question 3: When Did Indonesia Experience a Major Decline in Unemployment, and When Did the Trend Reverse?

### Analysis Background

Unemployment trends in Indonesia do not always move in the same direction every year. In some periods, unemployment consistently decreases across multiple education levels, while in other periods the trend begins to slow down or rise again.

This analysis aims to identify the turning points in unemployment trends by observing the average unemployment rate across years.

```python
import pandas as pd

df = pd.read_csv('Data_Cleaning_Pengangguran.csv')
```

```python
df_yearly = df.groupby('Periode')['Total'].mean().reset_index()
```

```python
df_yearly['Perubahan'] = df_yearly['Total'].diff()
```

```python
df_yearly
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

plt.plot(
    df_yearly['Periode'],
    df_yearly['Total'],
    marker='o'
)

plt.title('Trend of Unemployment in Indonesia')
plt.xlabel('Year')
plt.ylabel('Average Total Unemployment')

plt.grid(True)

plt.show()
```

### **Question 3: When Did Indonesia Experience a Major Decline in Unemployment, and When Did the Trend Reverse?**

**Answer & Insights:**
Based on the time-series analysis from 2006 to 2022, Indonesia experienced its most significant and prolonged decline in unemployment between **2006 and 2014**, where total unemployment consistently dropped from roughly 11.01 million down to 7.19 million. This downward trend was briefly interrupted by a minor uptick in 2011 (+74,502) but achieved its historical lowest golden era around **2016 to 2019**, stabilizing at an all-time low of approximately 7.00 million unemployed people. The absolute sharpest single-year drop occurred in **2012**, where the total number of unemployed plunged by **-979,288 people** in just twelve months.

However, the trend experienced a massive and aggressive reversal starting in **2020**. Driven by the global COVID-19 pandemic shock, unemployment exploded by an unprecedented **+1,345,010 people** in 2020 alone, followed by another substantial increase of +577,410 in 2021, peaking near the 9.00 million mark. Although the market showed structural resilience and began expanding again in 2022 with a post-pandemic recovery decline of **-509,988 people**, total unemployment has not yet returned to its stable pre-pandemic levels.

## 🎯 Portfolio Summary: The Structural Reality of Indonesia's Labor Market (2006-2022)

**Overall Insights:**
When we connect the dots across all findings, the 16-year trajectory of Indonesia's labor market reveals a critical structural shift. On a macro level, the country successfully engineered a massive, decade-long reduction in total unemployment, shrinking the pool from over 11 million in 2006 to a stable, flat "golden era" of around 7 million between 2016 and 2019. Even when external shocks like the 2020 global pandemic disrupted this progress—causing an unprecedented explosion of over 1.3 million unemployed people in a single year—the market demonstrated structural resilience by pivoting into a post-pandemic recovery phase by 2022.

However, the deepest value of this analysis lies beneath the surface. While overall numbers were dropping or stabilizing, an alarming "Academic Paradox" was silently brewing. The data proves that higher education has become the fastest-growing contributor to unemployment, with University graduates facing an aggressive 5.85% average annual increase—drastically outpaceing vocational tracks (SMK at 2.70%) and general high schools (SMU at -0.72%). 

This contrast suggests that while the economy has successfully absorbed lower-to-middle skilled labor (often into a flexible informal sector), it faces a severe structural mismatch in absorbing high-skilled talent. The growth of high-value, tech-driven industries is not expanding fast enough to match the supply of new higher education graduates. For stakeholders, recruiters, and policymakers, this data underscores a vital takeaway: the future focus must pivot from simply "creating jobs" to aggressively bridging the skills gap through modern, industry-aligned curriculum and strategic professional upskilling.
